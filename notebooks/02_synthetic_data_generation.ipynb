```python
import pandas as pd
import numpy as np

# Make our random data reproducible
rng = np.random.default_rng(42)

print("Libraries imported successfully!")
```

```python
organisations = pd.DataFrame({
    "organisation_id": ["ORG001", "ORG002", "ORG003"],
    "organisation_name": [
        "Alpha Capital",
        "Beta Investments",
        "Gamma Wealth"
    ],
    "country": [
        "Nigeria",
        "Ghana",
        "Kenya"
    ],
    "created_at": pd.to_datetime([
        "2023-01-15",
        "2023-03-20",
        "2023-06-10"
    ])
})

organisations
```

```python
print("Number of organisations:", len(organisations))

print(
    "Unique organisation IDs:",
    organisations["organisation_id"].nunique()
)

print("\nData types:")
print(organisations.dtypes)
```

```python
first_names = [
    "John", "Sarah", "David", "Mary", "Daniel",
    "Grace", "Michael", "Faith", "James", "Esther",
    "Samuel", "Joy", "Peter", "Ruth", "Joseph"
]

last_names = [
    "Smith", "Johnson", "Brown", "Williams", "Jones",
    "Garcia", "Miller", "Davis", "Wilson", "Taylor",
    "Thomas", "Moore", "Martin", "Lee", "White"
]

print("Customer name lists created!")
```

```python
customer_records = []

for organisation_id in organisations["organisation_id"]:

    for i in range(5):

        customer_id = f"CUST{i+1:03d}"

        first_name = rng.choice(first_names)
        last_name = rng.choice(last_names)

        email = (
            f"{first_name.lower()}.{last_name.lower()}"
            f"@example.com"
        )

        registration_date = pd.Timestamp(
            "2023-01-01"
        ) + pd.Timedelta(days=int(rng.integers(1, 1000)))

        customer_records.append({
            "organisation_id": organisation_id,
            "customer_id": customer_id,
            "first_name": first_name,
            "last_name": last_name,
            "email": email,
            "registration_date": registration_date
        })

customers = pd.DataFrame(customer_records)

customers
```

```python
print("Total customers:", len(customers))

print(
    "\nCustomers per organisation:"
)

print(
    customers.groupby("organisation_id")["customer_id"].count()
)

print(
    "\nUnique composite customer keys:",
    customers[["organisation_id", "customer_id"]]
    .drop_duplicates()
    .shape[0]
)
```

```python
account_types = ["Savings", "Current", "Investment"]
account_statuses = ["Active", "Inactive", "Closed"]

print("Account categories created!")
```

```python
account_records = []

for _, customer in customers.iterrows():

    for i in range(2):

        account_records.append({
            "account_id": f"ACC{len(account_records) + 1:04d}",
            "organisation_id": customer["organisation_id"],
            "customer_id": customer["customer_id"],
            "account_type": rng.choice(account_types),
            "balance": round(float(rng.uniform(1000, 500000)), 2),
            "account_status": rng.choice(account_statuses),
            "opened_at": pd.Timestamp("2023-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 1000)))
        })

accounts = pd.DataFrame(account_records)

accounts.head()
```

```python
print("Total accounts:", len(accounts))

print("\nAccounts per organisation:")
print(accounts.groupby("organisation_id")["account_id"].count())

print(
    "\nUnique account IDs:",
    accounts["account_id"].nunique()
)

print(
    "\nAccounts without a matching customer:",
    len(
        accounts.merge(
            customers[["organisation_id", "customer_id"]],
            on=["organisation_id", "customer_id"],
            how="left",
            indicator=True
        ).query("_merge == 'left_only'")
    )
)
```

```python
advisor_first_names = [
    "Andrew", "Rebecca", "Charles", "Linda",
    "Matthew", "Helen", "Robert", "Angela"
]

advisor_last_names = [
    "Adams", "Baker", "Clark", "Evans",
    "Hall", "King", "Lewis", "Scott"
]

print("Advisor name lists created!")
```

```python
advisor_records = []

for organisation_id in organisations["organisation_id"]:

    for i in range(3):

        first_name = rng.choice(advisor_first_names)
        last_name = rng.choice(advisor_last_names)

        advisor_records.append({
            "advisor_id": f"ADV{len(advisor_records) + 1:03d}",
            "organisation_id": organisation_id,
            "advisor_name": f"{first_name} {last_name}",
            "email": (
                f"{first_name.lower()}.{last_name.lower()}"
                f"@example.com"
            ),
            "specialisation": rng.choice([
                "Wealth Management",
                "Investment Planning",
                "Retirement Planning"
            ]),
            "hire_date": pd.Timestamp("2020-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 2000)))
        })

advisors = pd.DataFrame(advisor_records)

advisors.head()
```

```python
print("Total advisors:", len(advisors))

print("\nAdvisors per organisation:")
print(advisors.groupby("organisation_id")["advisor_id"].count())

print(
    "\nUnique advisor IDs:",
    advisors["advisor_id"].nunique()
)

print(
    "\nAdvisors without a matching organisation:",
    len(
        advisors.merge(
            organisations[["organisation_id"]],
            on="organisation_id",
            how="left",
            indicator=True
        ).query("_merge == 'left_only'")
    )
)
```

```python
portfolio_names = [
    "Growth Portfolio",
    "Income Portfolio",
    "Balanced Portfolio",
    "Retirement Portfolio",
    "Wealth Preservation"
]

risk_levels = ["Low", "Medium", "High"]

print("Portfolio categories created!")
```

```python
portfolio_records = []

for _, customer in customers.iterrows():

    organisation_advisors = advisors[
        advisors["organisation_id"] == customer["organisation_id"]
    ]["advisor_id"].tolist()

    for i in range(2):

        portfolio_records.append({
            "portfolio_id": f"PORT{len(portfolio_records) + 1:04d}",
            "organisation_id": customer["organisation_id"],
            "customer_id": customer["customer_id"],
            "advisor_id": rng.choice(organisation_advisors),
            "portfolio_name": rng.choice(portfolio_names),
            "risk_level": rng.choice(risk_levels),
            "created_at": pd.Timestamp("2023-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 1000)))
        })

portfolios = pd.DataFrame(portfolio_records)

portfolios.head()
```

```python
print("Total portfolios:", len(portfolios))

print("\nPortfolios per organisation:")
print(portfolios.groupby("organisation_id")["portfolio_id"].count())

print(
    "\nUnique portfolio IDs:",
    portfolios["portfolio_id"].nunique()
)

print(
    "\nPortfolios without a matching customer:",
    len(
        portfolios.merge(
            customers[["organisation_id", "customer_id"]],
            on=["organisation_id", "customer_id"],
            how="left",
            indicator=True
        ).query("_merge == 'left_only'")
    )
)
```

```python
portfolio_advisor_check = portfolios.merge(
    advisors[["organisation_id", "advisor_id"]],
    on="advisor_id",
    how="left",
    suffixes=("_portfolio", "_advisor"),
    indicator=True
)

invalid_advisor_assignments = portfolio_advisor_check[
    (portfolio_advisor_check["_merge"] == "left_only") |
    (
        portfolio_advisor_check["organisation_id_portfolio"]
        != portfolio_advisor_check["organisation_id_advisor"]
    )
]

print(
    "Portfolios with invalid advisor assignments:",
    len(invalid_advisor_assignments)
)
```

```python
security_data = [
    ("SEC001", "Apple Inc.", "Stock"),
    ("SEC002", "Microsoft Corp.", "Stock"),
    ("SEC003", "Amazon.com Inc.", "Stock"),
    ("SEC004", "Tesla Inc.", "Stock"),
    ("SEC005", "Alphabet Inc.", "Stock"),
    ("SEC006", "US Treasury Bond", "Bond"),
    ("SEC007", "Corporate Bond A", "Bond"),
    ("SEC008", "Vanguard S&P 500 ETF", "ETF"),
    ("SEC009", "iShares Core MSCI ETF", "ETF"),
    ("SEC010", "Global Equity Fund", "Mutual Fund")
]

securities = pd.DataFrame(
    security_data,
    columns=["security_id", "security_name", "security_type"]
)

securities
```

```python
securities["market_price"] = [
    round(float(rng.uniform(20, 500)), 2)
    for _ in range(len(securities))
]

securities["currency"] = "USD"

securities
```

```python
print("Total securities:", len(securities))

print(
    "\nUnique security IDs:",
    securities["security_id"].nunique()
)

print("\nSecurities by type:")
print(securities["security_type"].value_counts())

print(
    "\nMissing security IDs:",
    securities["security_id"].isna().sum()
)
```

```python
holding_records = []

for _, portfolio in portfolios.iterrows():

    # Each portfolio holds 3 different securities
    selected_securities = rng.choice(
        securities["security_id"],
        size=3,
        replace=False
    )

    for security_id in selected_securities:

        holding_records.append({
            "holding_id": f"HOLD{len(holding_records) + 1:04d}",
            "portfolio_id": portfolio["portfolio_id"],
            "organisation_id": portfolio["organisation_id"],
            "security_id": security_id,
            "quantity": int(rng.integers(1, 101)),
            "purchase_price": round(float(rng.uniform(20, 500)), 2),
            "purchase_date": pd.Timestamp("2023-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 1000)))
        })

holdings = pd.DataFrame(holding_records)

holdings.head()
```

```python
print("Total holdings:", len(holdings))

print(
    "\nUnique holding IDs:",
    holdings["holding_id"].nunique()
)

print(
    "\nHoldings without a matching portfolio:",
    len(
        holdings.merge(
            portfolios[["portfolio_id"]],
            on="portfolio_id",
            how="left",
            indicator=True
        ).query("_merge == 'left_only'")
    )
)

print(
    "\nHoldings without a matching security:",
    len(
        holdings.merge(
            securities[["security_id"]],
            on="security_id",
            how="left",
            indicator=True
        ).query("_merge == 'left_only'")
    )
)
```

```python
transaction_types = ["Deposit", "Withdrawal", "Transfer"]
transaction_records = []

for _, account in accounts.iterrows():

    for i in range(5):

        transaction_records.append({
            "transaction_id": f"TXN{len(transaction_records) + 1:05d}",
            "organisation_id": account["organisation_id"],
            "account_id": account["account_id"],
            "transaction_type": rng.choice(transaction_types),
            "amount": round(float(rng.uniform(100, 50000)), 2),
            "transaction_date": pd.Timestamp("2024-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 900)))
        })

transactions = pd.DataFrame(transaction_records)

print("Total transactions:", len(transactions))
transactions.head()
```

```python
trade_types = ["BUY", "SELL"]
trade_records = []

for _, holding in holdings.iterrows():

    for i in range(2):

        trade_records.append({
            "trade_id": f"TRD{len(trade_records) + 1:05d}",
            "organisation_id": holding["organisation_id"],
            "portfolio_id": holding["portfolio_id"],
            "security_id": holding["security_id"],
            "trade_type": rng.choice(trade_types),
            "quantity": int(rng.integers(1, 51)),
            "trade_price": round(float(rng.uniform(20, 500)), 2),
            "trade_date": pd.Timestamp("2024-01-01")
            + pd.Timedelta(days=int(rng.integers(1, 900)))
        })

trades = pd.DataFrame(trade_records)

print("Total trades:", len(trades))
trades.head()
```

```python
fee_types = ["Management Fee", "Transaction Fee", "Advisory Fee"]

fee_records = []

for organisation_id in organisations["organisation_id"]:

    for fee_type in fee_types:

        fee_records.append({
            "fee_schedule_id": f"FEE{len(fee_records) + 1:03d}",
            "organisation_id": organisation_id,
            "fee_type": fee_type,
            "fee_percentage": round(float(rng.uniform(0.1, 3.0)), 2),
            "effective_date": pd.Timestamp("2024-01-01")
        })

fee_schedules = pd.DataFrame(fee_records)

print("Total fee schedules:", len(fee_schedules))
fee_schedules.head()
```
